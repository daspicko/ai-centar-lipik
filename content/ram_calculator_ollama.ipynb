RAM (GB) ≈ model_GB + (context / 1000 × model_B × 0.5)

```python
def racunaj_ram(model_size, context_size, model_B):
    return model_size + (context_size/1000 * model_B * 0.5)

def izracunaj_max_context(ram_size, model_size, model_b):
    return int(((ram_size - model_size) * 2 * 1000) / model_b)
```

```python
RAM_SIZE = 128 # GB
CONTEXT_SIZE = 40000 # tokens
MODEL_SIZE = 2.5 # GB
MODEL_B = 4 # Parameters

print("Potrebno RAM-a:", racunaj_ram(MODEL_SIZE, CONTEXT_SIZE, MODEL_B))
print("Mogući context:", izracunaj_max_context(RAM_SIZE, MODEL_SIZE, MODEL_B))
```
