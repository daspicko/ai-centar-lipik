```python
import numpy as np
```

### [Random sampling](https://numpy.org/doc/stable/reference/random/index.html)

```python
rng = np.random.default_rng(42) # https://numpy.org/doc/stable/reference/random/generator.html

rng.integers(1, 11, size=5) # https://numpy.org/doc/stable/reference/random/generated/numpy.random.Generator.integers.html
rng.random(size=5) # https://numpy.org/doc/stable/reference/random/generated/numpy
rng.choice(['apple', 'banana', 'cherry', 'pear'], size=3) # https://numpy.org/doc/stable/reference/random/generated/numpy.random.Generator.choice.html
```

### [Masked arrays operations](https://numpy.org/doc/stable/reference/routines.ma.html)

```python
vector = np.array([1, 2, 3, 4, 5])

np.mean(vector) # https://numpy.org/doc/stable/reference/generated/numpy.ma.mean.html
np.median(vector) # https://numpy.org/doc/stable/reference/generated/numpy.ma.median.html
np.std(vector) # https://numpy.org/doc/stable/reference/generated/numpy.std.html
```

#### [Reshape](https://numpy.org/doc/stable/reference/generated/numpy.ndarray.reshape.html)

```python
matrix_4x3= np.array([[1, 4, 5], [10, 13, 9], [10, 13, 9], [10, 8, 5]])
matrix_4x3.reshape(2, -1) # 2 Rows, -1 means infer the number of columns
```

```python
matrix_4x3.reshape(-1, 2) # 2 Columns, -1 means infer the number of rows
```

#### [Searching](https://numpy.org/doc/stable/reference/routines.sort.html)

```python
argmax_row = np.argmax(matrix_4x3, axis=1) # https://numpy.org/doc/stable/reference/generated/numpy.argmax.html
argmax_row
```

```python
max_row = np.max(matrix_4x3, axis=1) # https://numpy.org/doc/stable/reference/generated/numpy.max.html
max_row
```

```python
my_matrix = np.array([[1, 4, 5], [10, 13, 9], [10, 8, 5]])
```

```python
every_second_row = my_matrix[::2, :] # https://numpy.org/doc/stable/user/basics.indexing.html
every_second_row
```

```python
every_second_column_reversed = my_matrix[:, ::-2]
every_second_column_reversed
```

```python
even_numbers = my_matrix[my_matrix % 2 == 0] # https://numpy.org/doc/stable/reference/generated/numpy.ndarray.__mod__.html
even_numbers
```

```python
odd_numbers = my_matrix[my_matrix % 2 == 1]
odd_numbers
```
